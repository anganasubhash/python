```python
#Prime number check
#using function
#without arguemen
#type 1.
def prime():
    n=int(input("Enter the number:"))
    flag=0
    for i in range(2,n):
        if n%i==0:
            flag=1
            break
    if flag>0:
        print("Not a prime")
    else:
        print("Prime")
prime()
        



```

```python
#type2
n=int(input("Enter the number:"))
def prime(n):
    flag=0
    for i in range(2,n):
        if n%i==0:
            flag=1
            break
    if flag>0:
        print("Not a prime")
    else:
        print("prime")
prime(n)
n2=int(input("Enter the number2:"))
prime(n2)
```
