```python
#Mutable /unmutable
dict2={
    "Name":"Anjana",
    "id"  :101,
    "roll number":101,
    "Age"  :40,

}
print(dict2)
dict2["Age"]=34
print(dict2)
```

```python
#Mutable/unmutable
#mutable---->can change


dict={
    "id"   :101,
    "fname":"Ram",
    "l name":"k",
    "Age"   :45,
    "designation":"software engineer",
    "salary":300000
}
print(dict)
dict["designation"]="Tester"
print(dict)
```

```python
dict2={
      "id"  :101,
        "fname":"Ram",
        "l name":"k",
        "Age"   :45,
        "designation":"software engineer",
        "salary":300000
}

for i in dict2:
    print(i,end=', ')

#only  key  element print
#To  print values
#id
print(dict2["id"])
#Designation
print(dict2["designation"])
#Age
print(dict2["Age"])
#To print key and value at a time
for i in dict2:
    print(i,":",dict2[i])
```

```python
#sentence to dictonary
sentence="cat dog cat dog rat bat cat dog rat bat rat"
lst1=sentence.split( )
print(lst1)
#1 st to list
dict={}
for i in lst1:#cat
    if i not in dict:
        dict[i]=1
        #cat:1,dog:1,
    else:
        dict[i]+=1#dic[cat]+=1  #cat=2
print(dict)
```
