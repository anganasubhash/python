```python
#sentence to dictonary
sentence="cat dog cat dog rat bat cat dog rat bat rat"
lst1=sentence.split( )
print(lst1)
#1 st to list
dict={}
for i in lst1:#cat
    if i not in dict:
        dict[i]=1
        #cat:1,dog:1,
    else:
        dict[i]+=1#dic[cat]+=1  #cat=2
print(dict)
```

```python
dic12={
    "car":3000,
    "bike":2000,
    "bus" :7500,
    "lorry":10000,
    "jeep":7500
}
#Method
#---------------
#for key values------key()
print(dic12.keys())
#for  values
print(dic12.values())
#get
print(dic12.get("car"))
#update
dic12.update({"Train":50000,"Aerplane":10000})
print(dic12)
dic12.pop("car")
print(dic12)
dic12.popitem()
print(dic12)
```

```python
student={
    "Alice":60,
    "Boby" :56,
    "David" :78,
    "Ram" :89,
    "gopu":23,
    "priya":44
}
pass_student={}
fail_student={}
for key,value in student.items():
    if value>50:
        pass_student[key]=value
    else:
        fail_student[key]=value
print(pass_student)
print(fail_student)
```
