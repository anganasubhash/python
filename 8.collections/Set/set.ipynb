```python
#Homogenous/hetrogenous
#Hetrogenous---it allows different data type
set1={10,True,False,6.9,"Hello","jeep"}
print(set1)
```

```python
##duplicated allowed/not
set2={10,67,10,"jeep","car","jeep",True,False}
print(set2)
#Duplicate not allowed
```

```python
#mutable/unmutable
#add----->add one element
set123={10,20,45,34,23}
print(set123)
set123.add(56)
print(set123)
#update------->add more than one element
set123.update([23,"hello","jeep"])
print(set123)
#set is mutable

```

```python
set1={10,45,34,67,31,12}
#sum
print(sum(set1))
#Length
print(len(set1))
#maximum
print(max(set1))
#Minimum
print(min(set1))
```
