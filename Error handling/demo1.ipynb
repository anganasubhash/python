```python

try:
    n1=int(input("Enter the number1:"))
    n2=int(input("Enter the number2:"))
    res=n1/n2
    print(res)
except ValueError:
    print("Enter valid numerical")
except ZeroDivisionError:
    print("Second digit must be a value not 0")
finally:
    print("Thank you,claculator closed")
```

```python
#login 
#username,password---login sucessfull
#raise error---value error
#except exception 
#print(statement)
#finally
try:
    username=input("Enter your username;")
    password=input("Enter your password:")
    if username=="" or password=="":
        raise ValueError("username / password both empty")
    print("Login sucessfull")
except Exception as e:
    print(e)
finally:
    print("Finally block executed")
```
