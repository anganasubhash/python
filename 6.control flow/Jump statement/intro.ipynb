```python

#Break ----->Stop
#1 to 100
#i=50 stop

for i in range(1,101):
    if i==50:
        break
    print(i)
```

```python
n=int(input("Enter the number:"))
flag=0
for i in range(2,n):
    if n%i==0:
        flag=1
        break
if flag>0:
    print("Not prime")
else:
    print("Prime")


```

```python
#continue-----> skip the condition
#1 to 50
#25
for i in range(1,51):
    if i==25:
        continue
    print(i)
```
