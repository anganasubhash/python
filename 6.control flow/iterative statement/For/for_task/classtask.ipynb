```python
#q1)common divisors num = 8  
     #   o/p=> 1,2,4,8

num=8
for i in range(1,9):
   if  num%i==0:
      print(i,end=" ")
```

```python
#2)common divisors of two numbers num1=8 and num2=24 
    #    #o/p  1,2,4,8

num1=8
num2=24
for i in range(1,9):
    if num1%i==0 and num2%i==0:
        print(i,end=" ")
```

```python
#q3)gcd of two numbers num1=8 and num2=24
num1=8
num2=12
if num1<num2:
    small=num1
else:
    small=num2
gcd=0
for i in range(1,small+1):
    if num1%i==0 and num2%i==0:
       gcd=i
print(gcd)
       
```

```python
#4)write a program to print an year is leapyear or not
year=int(input("Enter the year:"))
if year%400==0 or(year%4==0 and year%100!=0):
    print(year,"is leap year")
else:
    print(year,"Not leap year")
```
